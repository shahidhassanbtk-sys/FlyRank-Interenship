## 3. Leakage audit

I performed a leakage audit on the final feature set before interpreting the model results.

The main checks were:

1. **Future information:** No feature should use information from the future outcome period.
2. **Label leakage:** The target/label itself must not be included as a feature.
3. **Post-outcome information:** Features created after the outcome period should not be used.
4. **Duplicate or identifier leakage:** IDs or fields that directly reveal the target should not be used as predictive features.
5. **Validation leakage:** The same client should not appear in both the training and test sets when using the grouped validation design.

The purpose of this audit is to make sure that the model is learning from information that would actually be available at prediction time.

Based on these checks, the final feature set should contain only information available before the target outcome was measured.

## 4. Claim rewrite

### Original bold claim

> The model can accurately identify which pages will perform poorly and should be prioritized for action.

### Safer rewritten claim

> In this dataset, the model **measured a directional relationship** between the available features and the observed outcome. The results can be used as **decision-support** for reviewing and prioritizing pages, but they do not prove that the model will accurately predict outcomes for every future client or dataset.

The rewritten claim uses careful language because the validation results are based on the data and validation design used in this notebook. The model provides evidence that can support decisions, rather than a guarantee of future performance.

```python
print("Original claim:")
print("The model can accurately identify which pages should be prioritized.")

print("\nSafe claim:")
print("The model showed an observed and measured directional relationship "
      "between the available features and the outcome in this dataset. "
      "It can be used as decision-support for prioritizing pages, "
      "but it should not be treated as a guarantee of future results.")
```
