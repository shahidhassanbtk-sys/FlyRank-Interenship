## 1. Ranked actions + reason codes

The goal of this section is to turn the model output into a practical action queue.

Each item is ranked using the model's predicted score. A reason code is then assigned to explain, in simple human-readable words, why the item appears in the queue.

The reason codes are intended to support review and decision-making. They are not guarantees that an action will improve the outcome.

The main action categories are:

- **HIGH_PRIORITY_REVIEW** — high model score and should be reviewed first.
- **LOW_VISIBILITY_REVIEW** — the item has a visibility-related signal that may need investigation.
- **LOW_ENGAGEMENT_REVIEW** — the item has a weak engagement signal and may need content review.
- **GENERAL_REVIEW** — the item is flagged by the model but does not match a specific reason code.

The final queue is ranked from the highest model score to the lowest score.

## 2. Intended use and limits

### Intended use

This action playbook is designed for content and SEO teams to use as a decision-support tool.

The output can be used to:

The reason codes describe observed signals in the data. They do not prove the cause of an outcome.

Therefore, the playbook should be used for **decision-support and prioritization**, not as a replacement for human judgment or as a guarantee of future performance.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## 3. Human review + the no-go list

### Human review

Before taking action, a person should check:

- the page and its current content
- the model score and reason code
- the available search and performance data
- whether the recommendation makes sense for the page
- whether the page has enough data to support the decision
- whether the proposed action could have unintended effects

The model output is a prioritization tool. A human should review the evidence before making changes.

### No-go list

The following should never be fully automated:

- publishing or deleting content without human review
- making major content changes only because of a model score
- treating a reason code as proof of the cause
- applying the same action to every page without checking context
- making decisions when important data is missing or unreliable
- using the model outside the type of data and problem it was validated on

A zero-impression page should not automatically be sent through a content-refresh workflow because zero impressions can indicate an indexing or discoverability problem that belongs to a different pipeline.

The final decision should remain with a human reviewer.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ML-10 — Section 3: Human review + no-go checks

print("HUMAN REVIEW CHECKLIST")
print("=" * 60)

review_checks = [
    "Check the page and its current content",
    "Check the model score and reason code",
    "Check available search and performance data",
    "Confirm the recommendation makes sense for the page",
    "Check whether enough data is available",
    "Consider possible unintended effects"
]

for i, check in enumerate(review_checks, 1):
    print(f"{i}. {check}")

print("\nNO-GO LIST")
print("=" * 60)

no_go_rules = [
    "Do not publish or delete content without human review",
    "Do not make major content changes only because of a model score",
    "Do not treat a reason code as proof of the cause",
    "Do not apply the same action to every page without checking context",
    "Do not act when important data is missing or unreliable",
    "Do not use the model outside the data/problem it was validated on",
    "Do not automatically send zero-impression pages to a content-refresh workflow"
]

for i, rule in enumerate(no_go_rules, 1):
    print(f"{i}. {rule}")

print("\nFINAL RULE")
print("=" * 60)
print("The model supports prioritization; the final decision remains with a human reviewer.")
```

## 4. Monitoring / retrain triggers

Recommendations can become stale when the data or model performance changes.

### Monitoring checks

We should monitor:

- model performance on new data
- changes in the distribution of important features
- changes in the number of high-priority recommendations
- changes in recommendation outcomes
- missing or unusual input data
- changes in client or website behavior

### Retrain triggers

Consider retraining the model when:

- model performance drops on new validation data
- feature distributions change substantially
- recommendation quality becomes consistently worse
- new types of pages or clients appear
- important data sources or features change

The model should be reviewed before retraining. Retraining should use appropriate validation so that the updated model is checked on data it has not seen.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ML-10 — Section 4: Monitoring / retrain triggers

print("MONITORING CHECKS")
print("=" * 60)

monitoring_checks = [
    "Check model performance on new data",
    "Check changes in important feature distributions",
    "Check changes in the number of high-priority recommendations",
    "Check recommendation outcomes",
    "Check for missing or unusual input data",
    "Check for changes in client or website behavior"
]

for i, check in enumerate(monitoring_checks, 1):
    print(f"{i}. {check}")

print("\nRETRAIN TRIGGERS")
print("=" * 60)

retrain_triggers = [
    "Model performance drops on new validation data",
    "Important feature distributions change substantially",
    "Recommendation quality becomes consistently worse",
    "New types of pages or clients appear",
    "Important data sources or features change"
]

for i, trigger in enumerate(retrain_triggers, 1):
    print(f"{i}. {trigger}")

print("\nFINAL RULE")
print("=" * 60)
print("Review the model before retraining and validate the updated model on unseen data.")
```

```python
# ML-10 — Section 5: Export for the paper

from pathlib import Path
import pandas as pd

# Create output folder
output_dir = Path("work/outputs")
output_dir.mkdir(parents=True, exist_ok=True)

# Find all DataFrames currently available
dataframes = {
    name: value
    for name, value in globals().items()
    if isinstance(value, pd.DataFrame)
}

print("Available DataFrames:")
print("=" * 60)

if len(dataframes) == 0:
    print("No DataFrame is currently available.")
    print("\nPlease run the previous notebook cells first.")
else:
    for name, df in dataframes.items():
        print(f"{name}: {df.shape}")
        print("Columns:", list(df.columns))
        print()

# Look for a DataFrame that contains a score column
score_names = [
    "model_score",
    "score",
    "action_score",
    "baseline_score",
    "predicted_score"
]

selected_df = None
selected_name = None

for name, df in dataframes.items():
    for score_col in score_names:
        if score_col in df.columns:
            selected_df = df.copy()
            selected_name = name
            selected_score_col = score_col
            break
    if selected_df is not None:
        break

# If a suitable DataFrame was found
if selected_df is not None:

    # Rank by the score
    selected_df = selected_df.sort_values(
        by=selected_score_col,
        ascending=False
    ).reset_index(drop=True)

    # Add rank
    selected_df["rank"] = selected_df.index + 1

    # Save CSV
    queue_path = output_dir / "action_queue.csv"
    selected_df.to_csv(queue_path, index=False)

    print("=" * 60)
    print("EXPORT SUCCESSFUL")
    print("=" * 60)
    print("DataFrame used:", selected_name)
    print("Score column:", selected_score_col)
    print("Rows:", len(selected_df))
    print("Saved to:", queue_path)

    # Show first 20 rows
    display(selected_df.head(20))

else:
    print("=" * 60)
    print("NO SCORE DATAFRAME FOUND")
    print("=" * 60)
    print("Run the previous ML-10/model cells first.")
    print("The code needs a DataFrame containing a score column.")
```
