## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/shahidhassanbtk-sys/FlyRank-Interenship"
REPO_DIR = "FlyRank-Interenship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

print("Current working directory:", os.getcwd())
print("Rows:", df.shape[0], "Columns:", df.shape[1])
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
```

## 5. Why ML beats a fixed rule here

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Check: does the hand rule's cutoff actually separate declining vs stable pages cleanly?
import numpy as np

near_threshold = df[(df["days_since_last_update"].between(150, 210))]
print("Pages near the 180-day cutoff:", near_threshold.shape[0])
print("Declining rate just below/above cutoff:",
      round(near_threshold["is_declining_label"].mean(), 3))
print("Overall declining rate:", round(df["is_declining_label"].mean(), 3))
```
