## 3. Methodology

### Assumptions

The analysis assumes that:

- the available search and page-performance data is suitable for identifying pages that need review
- pages need enough history to measure a meaningful change in impressions
- the model is used for prioritization, not automatic action
- pages from the same client should not be mixed between training and testing

### Features

The model uses available page-level search and performance signals, such as:

- impressions
- clicks
- search volume
- average position
- content-related features
- page age and other available page signals

Features are selected from information that would be available for the page at the time of review.

### Label Definition

The target label represents whether a page is showing a decline in search visibility.

A page is treated as a positive example when its impressions have declined compared with the previous measurement period.

Pages with zero impressions are excluded from this pipeline because a declining-impressions trend cannot be measured reliably for them. They belong to a different indexing/discoverability problem.

### Baseline

The baseline is the existing hand-written rule.

The machine-learning model is compared with this fixed rule using the same evaluation data.

The comparison helps determine whether the learned model provides useful prioritization beyond the simple baseline.

### Validation Design

The validation uses a client-aware holdout split.

Pages belonging to a client are kept together, so the same client does not appear in both the training and test sets.

The model is trained only on the training clients and evaluated on unseen clients.

The hand rule is applied to the same held-out test set so that the comparison is fair.

### Leakage Checks

Potential leakage is checked before modeling.

In particular:

- information from the target/future period is not used as an input feature
- the target is not included among the model features
- train and test data are separated by client
- CTR is not used when it would leak the outcome because CTR is calculated from clicks and impressions
- preprocessing and model fitting are performed using training data only where applicable

These checks help prevent the model from receiving information that would not be available when making a real prediction.

## 6. Ranked recommendations

The model output is used to create a ranked action queue. Pages with higher scores are reviewed first.

### Recommendation order

1. **Review the highest-scoring pages first**
   - Start with the pages at the top of the ranked queue.
   - Check the model score and reason code.

2. **Investigate the supporting signals**
   - Review search, content, and performance data.
   - Check whether the observed signals support further investigation.

3. **Prioritize pages with clear evidence**
   - Give human reviewers priority to pages where the available data provides enough evidence for review.
   - Do not treat the score as proof that an action is required.

4. **Handle zero-impression pages separately**
   - Do not automatically send zero-impression pages to the content-refresh workflow.
   - Investigate indexing or discoverability issues through the appropriate pipeline.

5. **Human review before action**
   - A reviewer should inspect the page and evidence before making changes.
   - The final action remains a human decision.

### Recommendation Output

The final output is a ranked queue containing the page, its model score, and available reason codes.

The queue is intended to help teams decide **what to review first**, not to automatically decide **what action to take**.
