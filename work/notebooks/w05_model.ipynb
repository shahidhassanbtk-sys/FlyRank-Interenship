## 4. Errors and interpretation

The model can make two main types of errors:

- **False Positive:** The model predicts that an item will go dark, but it actually does not.
- **False Negative:** The model predicts that an item will not go dark, but it actually does.

For this task, false negatives are important because they may cause potentially declining content to be missed during review.

The model may lean on features that show recent content performance, such as impressions, clicks, CTR, and average position. Higher recent performance may reduce the predicted risk, while weaker performance may increase it.

Error analysis should focus on which items were misclassified and whether the model is missing important declining items.

```python
import pandas as pd

# Check whether the model is a Logistic Regression model
if hasattr(model, "coef_"):

    # Get feature names directly from the model
    if hasattr(model, "feature_names_in_"):
        feature_names = model.feature_names_in_

    else:
        # If feature names are not stored, use the number of model features
        feature_names = [
            f"Feature_{i+1}"
            for i in range(model.coef_.shape[1])
        ]

    feature_importance = pd.DataFrame({
        "feature": feature_names,
        "coefficient": model.coef_[0]
    })

    feature_importance["absolute_coefficient"] = (
        feature_importance["coefficient"].abs()
    )

    feature_importance = feature_importance.sort_values(
        "absolute_coefficient",
        ascending=False
    )

    print("Features the model leans on most:")
    display(feature_importance[["feature", "coefficient"]])

else:
    print("The current model does not provide Logistic Regression coefficients.")
```
